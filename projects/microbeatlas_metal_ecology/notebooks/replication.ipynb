# Independent Replication Hub

This notebook is the entry point for all independent dataset replication attempts.
Each sub-notebook tests whether the primary finding — metal gene density (per Mb) predicts
niche specialization — holds in an independent dataset or under a restricted analysis.

**Primary finding (MicrobeAtlas, n=997):** Total 94-KO/Mb β=−0.022, p=4×10⁻⁷;
Tier 2 homeostasis/Mb β=−0.009, p=0.011 (Tier 1 p=0.256, null).
Niche metric: Levins' B_std across all 13 Env_Level_1 categories.


## Replication Strategy

A true replication requires: (1) **independent genome source** (not MicrobeAtlas),
(2) **independent niche metric** (not the same OTU presence/absence matrix), and
(3) the **same per-Mb predictor** from the 94-KO tiered list.

We distinguish two types of replication here:
- **Strong replication**: different genomes + different community matrix (e.g., MGnify MAGs + MGnify biome)
- **Sensitivity replication**: same community matrix, restricted subset (e.g., soil-only Levins' B)
- **Independent dataset**: different amplicon survey entirely (e.g., AusMicrobiome, NETL)

**MGnify sign note:** MGnify uses biome Shannon H (cross-biome cosmopolitanism, broader = higher H),
while primary uses Levins' B within MicrobeAtlas (within-habitat specialization, narrower = lower B).
These measure orthogonal niche dimensions — both findings are biologically coherent.

**Soil-only result (NB14):** Signal is not driven by soil vs. non-soil contrasts;
β magnitude essentially unchanged (−0.023 vs. −0.022 primary), confirming within-soil
ecological differentiation.

**NB13 null:** AusMicrobiome uses vegetation-based ecological zones (temperate grassland, subtropical
shrubland), not metal-geochemical gradients. Absence of signal is expected given the niche axis.

**NB15 null (uninformative):** Three structural problems: (1) RDP taxonomy bridge loses ~85% of genera
before merging with GTDB metal data; (2) only 3 well-type categories give near-zero niche variance —
ABR negative control also significant (p=0.027), indicating a generic "gene-rich generalist" confound;
(3) well types classify hydrocarbon source formation, not metal exposure gradient. Ba-concentration
niche (Block 6) is underpowered (n=73) with λ=−0.75 (model misspecified).

## Interpretation of concordant and discordant results

| Pattern | Meaning |
|---|---|
| Same sign, same metric | Direct replication |
| Opposite sign, different metric | Orthogonal dimensions (expected: biome_H vs. Levins' B) |
| Null in independent dataset | Underpowered or dataset-specific (check n and env diversity) |
| Same sign, different metric | Robust generalisation across niche frameworks |

The discordance between MGnify (positive β, biome_H) and primary (negative β, Levins' B)
was confirmed by Diagnostic Tests A/B/C in `scripts/cross_dataset_validation.py`:
the sign reversal is driven by the niche metric, not the annotation source.
